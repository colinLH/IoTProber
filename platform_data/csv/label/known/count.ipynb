```python
import pandas as pd

devices = ["NAS", "NVR", "POWER_METER", "BUILDING_AUTOMATION", "MEDICAL", "ROUTER", "PRINTER", "SCADA", "CAMERA"]

for dev in devices:
    df = pd.read_csv(f"label_nonempty_{dev}.csv")
    vendor_counts = df['vendor'].value_counts()
    print(f"\n{dev} - Vendor distribution:")
    
    total_vendors = 0
    
    for vendor, count in vendor_counts.items():
        print(f"  {vendor}: {count}")
        
        total_vendors += 1
    
    print(f"{dev}: {total_vendors}")
    print(f"{dev}: {df.shape}")

```

```python
import pandas as pd

devices = ["NAS", "NVR", "POWER_METER", "BUILDING_AUTOMATION", "MEDICAL", "ROUTER", "PRINTER", "SCADA", "CAMERA", "MEDIA_SERVER", "VPN"]

non_empty = 0
for dev in devices:
    df = pd.read_csv(f"label_nonempty_{dev}.csv")
    print(f"{dev}: {df.shape}")
    non_empty += df.shape[0]
print(f"Total non-empty: {non_empty}")
```
